```python
# Gradient Decent
```

```python
import numpy as np
from sklearn.datasets import make_regression
```

```python
X,y = make_regression(n_samples=4, n_features=1,n_informative=1,n_targets=1,noise=80)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
lr.fit(X,y)
```

```python
import matplotlib.pyplot as plt

plt.scatter(X,y)
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
plt.scatter(X,y)
plt.plot(X, lr.predict(X), color='red')
```

```python
# let's apply Gradient decent assuming slope is constant 100.79
# and let's assume the starting value of b is equal to zero b = 0
y_pred = ((100.79 * X) + 100).reshape(4)
```

```python
plt.scatter(X,y)
plt.plot(X, lr.predict(X), color='red', label='OLS')
plt.plot(X,y_pred, color='#00a65a', label='b=0')
plt.legend()
plt.show()
```

```python
m = 100.79
b = 80

loss_slope = -2 * np.sum(y  - m*X.ravel() - b)
loss_slope
```

```python
# let's take learning rate = 0.01
LR = 0.1

step_size = loss_slope * LR
step_size
```

```python
b = b - step_size
b
```

```python
plt.scatter(X,y)
plt.plot(X,lr.predict(X), color='red', label='OLS')
plt.plot(X, y_pred, color='#00a65a', label='b = {}.format(b)')
plt.plot(X,y_pred, color="#209D84", label='b=0')
plt.legend()
plt.show()
```

```python
loss_slope = -2 * np.sum(y - m*X.ravel() -b)
loss_slope
```

```python
step_size = loss_slope*LR
step_size
```

```python
b = b-step_size
b
```

```python
y_pred2 = ((100.79*X)+b).reshape(4)
```

```python
plt.scatter(X,y)
plt.plot(X, lr.predict(X), color='red', label='OLS')
plt.plot(X,y_pred2,color='#A3E4D7',label='b = {}'.format(b))
plt.plot(X, y_pred2, color='#A3E4D7' ,label='b=0'.format(b))
plt.legend()
plt.show()
```
